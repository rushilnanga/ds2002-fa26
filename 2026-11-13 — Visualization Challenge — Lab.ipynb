```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BLUE, RUST, GRAY = '#4a5d7e', '#b4553f', '#c9d0d9'
sns.set_theme(style='white')
plt.rcParams['text.parse_math'] = False

CRITERIA = [
    'title states a finding',
    'axes labeled with units',
    'right chart for the question',
    'honest scale',
    'readable',
    'no wasted ink',
]

def score(**passed):
    """score(title=False, axes=True, ...) -> prints a scored critique."""
    total = 0
    for name, ok in zip(CRITERIA, passed.values()):
        print(('PASS' if ok else 'FAIL'), '-', name)
        total += bool(ok)
    print(f'\nscore: {total}/6')
    return total


def finish(ax, title, xlabel=None, ylabel=None, fontsize=12):
    """Same helper as Monday's lecture: title, labels, spines, tight_layout."""
    ax.set_title(title, loc='left', fontsize=fontsize)
    if xlabel is not None:
        ax.set_xlabel(xlabel)
    if ylabel is not None:
        ax.set_ylabel(ylabel)
    for spine in ('top', 'right'):
        ax.spines[spine].set_visible(False)
    ax.figure.tight_layout()


# --- game-day sample (same generator as the Monday lecture) ------------------
rng = np.random.default_rng(2026)
hours = np.arange(9, 22)
build_up = np.exp(-((hours - 14.5) ** 2) / (2 * 2.3 ** 2))
storm = np.exp(-((hours - 17.0) ** 2) / (2 * 1.5 ** 2))
GAMES = {'Sep 5': False, 'Sep 19': False, 'Oct 10': True, 'Nov 14': True}

rows = []
for game, rainy in GAMES.items():
    rain_mm = rainy * (0.4 + 3.0 * storm) * rng.gamma(10, 0.1, len(hours))
    for h, b, rain in zip(hours, build_up, rain_mm):
        n_main = rng.poisson(b * 150 + 12)
        n_rain = rng.poisson(rain * 5 + 0.2)
        cats = np.concatenate([
            rng.choice(['Food', 'Merch', 'Drink'], n_main, p=[.45, .22, .33]),
            np.repeat('RainGear', n_rain),
        ])
        rows += [(game, 'rain' if rainy else 'dry', h, round(rain, 1), c) for c in cats]

df = pd.DataFrame(rows, columns=['game', 'weather', 'hour', 'rain_mm', 'category'])
df['zone'] = rng.choice(['A', 'B', 'C'], len(df), p=[.6, .3, .1])
df['revenue'] = (
    df['category'].map({'Food': 12, 'Merch': 28, 'Drink': 6, 'RainGear': 15})
    * rng.lognormal(-0.18, 0.6, len(df))
).round(2)

print(len(df), 'orders across', df['game'].nunique(), 'games')
df.head()
```

### Chart 1 — revenue by category

Sorted horizontal bar. Axis from zero.

```python
# TODO

revenue_cat = df.groupby('category')['revenue'].sum().sort_values()

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.barh(revenue_cat.index, revenue_cat.values, color=BLUE)

ax.set_xlim(left=0)
ax.bar_label(bars, fmt='$%.0f', padding=4)
ax.set_xlim(0, revenue_cat.max() * 1.2)

finish(ax,
       'Merchandise Generates The Highest Gameday Revenue',
       xlabel='Total revenue (USD)',
       ylabel='Product category')

plt.show()
```

**Caption:** Merch generates the most revenue suggesting that it should be prioritized on gameday by disgnating extra staff and resources.

### Chart 2 — revenue by hour

Line chart, kickoff marked at 15.5, and the title should state the staffing implication rather than describing the axes.

```python
# TODO

hourly_revenue = df.groupby('hour')['revenue'].sum()

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(hourly_revenue.index, hourly_revenue.values,
        color=BLUE, marker='o', linewidth=2)

ax.axvline(15.5, color=RUST, linestyle='--',
           label='Kickoff (3:30 PM)')

ax.set_xlim(9, 21)
ax.set_ylim(bottom=0)
ax.set_xticks(range(9, 22, 2))
ax.legend(frameon=False)

finish(ax,
       'Staff More Heavily Before Kickoff When Revenue Peaks',
       xlabel='Hour of day (24-hour clock)',
       ylabel='Total revenue (USD)')

plt.show()
```

**Caption:** Revenue is highest in the hours leading up to kickoff, showcasing a need for increased staffing before the game begins

### Chart 3 — revenue by zone, normalized

Raw revenue by zone mostly measures how big each zone is. Use the capacity figures below to build a second view — revenue per unit of capacity — and show both side by side.

Then say which of the two you would put in front of a manager, and for which question.

```python
zone_capacity = {'A': 12000, 'B': 4000, 'C': 1500}
# TODO: two panels -- raw revenue and revenue per capacity

zone_revenue = df.groupby('zone')['revenue'].sum()
revenue_per_capacity = zone_revenue / pd.Series(zone_capacity)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

bars1 = axes[0].bar(zone_revenue.index, zone_revenue.values, color=BLUE)
axes[0].bar_label(bars1, fmt='$%.0f', padding=3)
axes[0].set_ylim(0, zone_revenue.max() * 1.2)
finish(axes[0],
       'Zone A Generates The Most Total Revenue',
       xlabel='Zone',
       ylabel='Total revenue (USD)')

bars2 = axes[1].bar(revenue_per_capacity.index,
                    revenue_per_capacity.values, color=RUST)
axes[1].bar_label(bars2, fmt='$%.2f', padding=3)
axes[1].set_ylim(0, revenue_per_capacity.max() * 1.2)
finish(axes[1],
       'Zone C Leads in Revenue per Capacity',
       xlabel='Zone',
       ylabel='Revenue per capacity unit (USD)')

plt.show()
```

**Caption, and which panel I would present:** Zone A generates the most total revenue while Zone C generates the most revenue relative to its capacity, supporting different staffing and resource allocation decisions. I would present the second chart to a manager evaluating zone efficiency because it accounts for differences in zone size.

### Chart 5 — your choice

**TODO:** a grouped, stacked, or scatter view that reveals something the first four do not. The finding is the point; the chart type is your call.

If you use a scatter, size or annotate the points so a reader can tell which ones carry enough data to trust.

```python
# TODO

orders_weather = (
    df.groupby(['weather', 'category'])
      .size()
      .unstack('weather', fill_value=0)
)

game_counts = df.groupby('weather')['game'].nunique()
orders_weather = orders_weather.div(game_counts, axis=1)

orders_weather = orders_weather[['dry', 'rain']]

fig, ax = plt.subplots(figsize=(9, 4))

orders_weather.plot(kind='bar', ax=ax,
                    color=[BLUE, RUST], width=0.7)

ax.set_ylim(bottom=0)
ax.set_xticklabels(orders_weather.index, rotation=0)
ax.legend(['Dry games', 'Rainy games'],
          title='Weather', frameon=False)

finish(ax,
       'Rainy Games Increase Demand For Rain Gear',
       xlabel='Product category',
       ylabel='Average orders per game')

plt.show()
```

**Caption:** While drink, food, and merch purchasese stay relatively similar, rainy games generate greater demand for rain gear helping managers adjust product inventory according to expected weather conditions
